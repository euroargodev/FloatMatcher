# Points

`set_points` takes arrays, a DataFrame, a Dataset or a PointSet: the loader is chosen from the type.

```python
import numpy as np
import pandas as pd
import xarray as xr

from floatmatcher.floatmatcher import FloatMatcher
from floatmatcher.pointset import PointSet

my_fm = FloatMatcher()
```

### Arrays

```python
lon = np.array([-28.15, -46.51])
lat = np.array([54.48, 30.08])
time = np.array(["2013-03-18T22:47", "2014-05-18T12:22"], dtype="datetime64[ns]")

my_fm.set_points(lon, lat, time)
my_fm.points
```

```python
# without time, for a grid without time axis (lut)
my_fm.set_points(lon, lat)
my_fm.points
```

### DataFrame

```python
df = pd.DataFrame({
    "longitude": [-38.2, -35.6, -31.9],
    "latitude":  [31.5, 34.2, 37.8],
    "date": pd.to_datetime(["2018-01-02 03:10", "2018-01-03 14:45", "2018-01-05 08:20"]),
})

my_fm.set_points(df)  # longitude, latitude, date by default
my_fm.points
```

```python
# other column names
df_custom = df.rename(columns={"longitude": "LON", "latitude": "LAT", "date": "DATE"})

my_fm.set_points(df_custom, lon="LON", lat="LAT", time="DATE")
my_fm.points
```

```python
# no time column: time=None (without it, a warning tells the column is missing)
my_fm.set_points(df.drop(columns="date"), time=None)
my_fm.points
```

### Dataset

```python
ds = xr.Dataset(
    {
        "LONGITUDE": ("N_PROF", [-40.0, -30.0, -20.0]),
        "LATITUDE": ("N_PROF", [35.0, 36.0, 37.0]),
        "TIME": ("N_PROF", pd.to_datetime(["2018-01-03", "2018-01-04", "2018-01-05"])),
    },
    coords={"N_PROF": [10, 11, 12]},
)

my_fm.set_points(ds)  # Argo names by default: LONGITUDE, LATITUDE, TIME
my_fm.points
```

### Argo float with argopy

```python
from argopy import DataFetcher

argo = DataFetcher().float(6902746).to_xarray().argo.point2profile()  # one point per profile

my_fm.set_points(argo)
my_fm.points
```

### PointSet

```python
points = PointSet.from_arrays(lon, lat, time)

my_fm.set_points(points)
my_fm.points
```

`reinject()` gives back the same type as the input: a DataFrame for a DataFrame, a Dataset for a Dataset. Arrays and PointSet only give `result`.
